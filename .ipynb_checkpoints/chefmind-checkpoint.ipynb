or the first improved experiment, we'll use potential-based reward shaping based on hand size.
The shaping idea is:
\[
**F(s,s') = \gamma \Phi(s')-\Phi(s)**
\]
where the potential is based on how many cards remain in the hand.

```python
import numpy as np
from agents.agent_dqn import DQNAgent


class ChefMindDQNAgent(DQNAgent):
    """
    Custom DQN agent for the ChefMind delayed-reward experiment.

    Adds potential-based reward shaping using the number
    of cards remaining in the player's hand.
    """

    def _get_potential(self, state):
        """
        Potential function:
        fewer cards in hand = better state.

        First 17 values of the state represent the hand.
        """
        hand = state[:17]

        # Count non-zero card entries
        cards_remaining = np.count_nonzero(hand)

        # Convert to a positive progress value
        return -float(cards_remaining)

    def _get_shaped_reward(self, old_state, new_state, base_reward):
        """
        Potential-based reward shaping:
        F(s,s') = gamma * Phi(s') - Phi(s)
        """
        old_potential = self._get_potential(old_state)
        new_potential = self._get_potential(new_state)

        shaping_reward = (
            self.gamma * new_potential - old_potential
        )

        return base_reward + 0.05 * shaping_reward
```

```python
# ============================================================
# CHEFMIND-RL — COMPLETE REMAINING EXPERIMENT
# Sparse / Delayed Reward Variant
# ============================================================

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from agents.agent_dqn import DQNAgent
from rooms.room import Room


# ============================================================
# 1. CUSTOM CHEFMIND DQN WITH POTENTIAL-BASED REWARD SHAPING
# ============================================================

class ChefMindDQNAgent(DQNAgent):
    """
    Custom Dueling DQN for the ChefMind delayed-reward experiment.

    The official final reward is retained.
    A small potential-based shaping signal is added
    according to progress in reducing cards in hand.
    """

    def _get_potential(self, state):
        """
        State structure:
        first 17 values = hand
        next 11 values = board

        Fewer cards remaining means greater progress.
        """
        hand = state[:17]

        # Count non-zero entries in the hand
        cards_remaining = np.count_nonzero(hand)

        # Smaller hand = better state
        return -float(cards_remaining)

    def _get_shaped_reward(self, old_state, new_state, base_reward):
        """
        Potential-based shaping:

        F(s,s') = gamma * Phi(s') - Phi(s)

        A small coefficient prevents shaping from
        dominating the original reward.
        """
        old_potential = self._get_potential(old_state)
        new_potential = self._get_potential(new_state)

        shaping = (
            self.gamma * new_potential
            - old_potential
        )

        return base_reward + 0.05 * shaping

    def request_action(self, observations):

        # Convert observation into the same 28-dimensional
        # state representation used by the official DQN.
        hand = np.array(
            observations["hand"]
        ).flatten() / 13

        board = np.array(
            observations["board"]
        ).flatten() / 13

        possible_actions_values = list(
            observations["possible_actions"]
        )

        obs = np.concatenate([hand, board])

        # Mask illegal actions
        possible_actions_mask = np.zeros(
            self.action_size,
            dtype=np.float32
        )

        valid_action_indices = [
            self.all_actions.index(val)
            for val in possible_actions_values
        ]

        possible_actions_mask[
            valid_action_indices
        ] = 1.0

        # Select action
        action_index = self.act(
            obs,
            possible_actions_mask,
            valid_action_indices
        )

        action_str = self.all_actions[action_index]

        # ----------------------------------------------------
        # Original reward shaping from official DQN
        # ----------------------------------------------------

        base_reward = 0.0

        if action_str.lower() == "pass":
            base_reward -= 1.0

        base_reward -= 0.02

        # ----------------------------------------------------
        # Additional potential-based shaping
        # ----------------------------------------------------

        if (
            self.last_state is not None
            and self.last_action is not None
            and self.train
            and self.last_possible_actions is not None
        ):

            shaped_reward = self._get_shaped_reward(
                self.last_state,
                obs,
                base_reward
            )

            self.episode.append(
                (
                    self.last_state,
                    self.last_possible_actions,
                    self.last_action,
                    shaped_reward,
                    obs,
                    possible_actions_mask,
                    False,
                )
            )

        # Store current transition information
        self.last_state = obs
        self.last_action = action_index
        self.last_possible_actions = possible_actions_mask

        return action_index


# ============================================================
# 2. CREATE FRESH SHAPED-REWARD ROOM
# ============================================================

output_folder = "/kaggle/working/chefmind_shaped_final"

os.makedirs(
    output_folder,
    exist_ok=True
)

shaped_room = Room(
    run_remote_room=False,
    room_name="ChefMind_Shaped_Final",
    max_matches=10,
    output_folder=output_folder,
    save_game_dataset=False,
    save_logs_game=False,
    save_logs_room=False,
)


# ============================================================
# 3. CREATE FOUR SHAPED DQN AGENTS
# ============================================================

shaped_agents = []

for i in range(4):

    agent = ChefMindDQNAgent(
        name=f"ShapedDQN{i}",
        log_directory=shaped_room.room_dir,
        train=True,
        verbose_console=False
    )

    shaped_agents.append(agent)

    shaped_room.connect_player(agent)


print("=" * 70)
print("CHEFMIND SHAPED-REWARD EXPERIMENT")
print("=" * 70)

print("Agents:", len(shaped_agents))
print("Matches:", 10)
print("Training: True")
print("Reward method: Potential-based shaping")
print("State size: 28")
print("Action size: 200")
print("Gamma:", shaped_agents[0].gamma)


# ============================================================
# 4. RUN TRAINING
# ============================================================

print("\nStarting shaped-reward training...")
print("-" * 70)

await shaped_room.run()

print("-" * 70)
print("Shaped-reward training completed!")


# ============================================================
# 5. EXTRACT SHAPED RESULTS
# ============================================================

shaped_results = []

for agent in shaped_agents:

    rewards = agent.rewards
    positions = agent.positions

    wins = sum(
        1 for p in positions
        if p == 1
    )

    win_rate = (
        wins / len(positions) * 100
        if positions
        else 0
    )

    average_position = (
        np.mean(positions)
        if positions
        else np.nan
    )

    average_reward = (
        np.mean(rewards)
        if rewards
        else np.nan
    )

    shaped_results.append({
        "Agent": agent.name,
        "Wins": wins,
        "Matches": len(positions),
        "Win Rate (%)": win_rate,
        "Average Position": average_position,
        "Average Reward": average_reward
    })


shaped_df = pd.DataFrame(shaped_results)


# ============================================================
# 6. ENTER YOUR ACTUAL BASELINE RESULTS
# ============================================================

baseline_data = [
    {
        "Agent": "DQN0",
        "Wins": 5,
        "Matches": 10,
        "Win Rate (%)": 50.0,
        "Average Position": 1.9,
        "Average Reward": 1.48
    },
    {
        "Agent": "DQN1",
        "Wins": 2,
        "Matches": 10,
        "Win Rate (%)": 20.0,
        "Average Position": 2.7,
        "Average Reward": 0.576
    },
    {
        "Agent": "DQN2",
        "Wins": 1,
        "Matches": 10,
        "Win Rate (%)": 10.0,
        "Average Position": 2.9,
        "Average Reward": 0.278
    },
    {
        "Agent": "DQN3",
        "Wins": 2,
        "Matches": 10,
        "Win Rate (%)": 20.0,
        "Average Position": 2.5,
        "Average Reward": 0.576
    }
]

baseline_df = pd.DataFrame(baseline_data)


# ============================================================
# 7. DISPLAY RESULTS
# ============================================================

print("\n")
print("=" * 70)
print("BASELINE DQN RESULTS")
print("=" * 70)

display(baseline_df)


print("\n")
print("=" * 70)
print("CHEFMIND SHAPED-DQN RESULTS")
print("=" * 70)

display(shaped_df)


# ============================================================
# 8. OVERALL COMPARISON
# ============================================================

baseline_overall = {
    "Experiment": "Baseline DQN",
    "Average Win Rate (%)":
        baseline_df["Win Rate (%)"].mean(),
    "Average Position":
        baseline_df["Average Position"].mean(),
    "Average Reward":
        baseline_df["Average Reward"].mean()
}

shaped_overall = {
    "Experiment": "ChefMind Shaped DQN",
    "Average Win Rate (%)":
        shaped_df["Win Rate (%)"].mean(),
    "Average Position":
        shaped_df["Average Position"].mean(),
    "Average Reward":
        shaped_df["Average Reward"].mean()
}

comparison_df = pd.DataFrame(
    [
        baseline_overall,
        shaped_overall
    ]
)


print("\n")
print("=" * 70)
print("BASELINE VS CHEFMIND")
print("=" * 70)

display(comparison_df)


# ============================================================
# 9. IMPROVEMENT CALCULATIONS
# ============================================================

baseline_win = baseline_overall[
    "Average Win Rate (%)"
]

shaped_win = shaped_overall[
    "Average Win Rate (%)"
]

baseline_position = baseline_overall[
    "Average Position"
]

shaped_position = shaped_overall[
    "Average Position"
]

baseline_reward = baseline_overall[
    "Average Reward"
]

shaped_reward = shaped_overall[
    "Average Reward"
]

print("\n")
print("=" * 70)
print("IMPROVEMENT")
print("=" * 70)

print(
    f"Win-rate change: "
    f"{shaped_win - baseline_win:.2f} percentage points"
)

print(
    f"Average-position change: "
    f"{shaped_position - baseline_position:.2f}"
)

print(
    f"Average-reward change: "
    f"{shaped_reward - baseline_reward:.3f}"
)


# ============================================================
# 10. PLOT WIN RATE
# ============================================================

plt.figure(figsize=(8, 5))

plt.bar(
    comparison_df["Experiment"],
    comparison_df["Average Win Rate (%)"]
)

plt.ylabel("Average Win Rate (%)")
plt.title("Baseline DQN vs ChefMind Shaped DQN")
plt.xticks(rotation=15)
plt.tight_layout()

win_plot = (
    "/kaggle/working/chefmind_win_rate_comparison.png"
)

plt.savefig(
    win_plot,
    dpi=200,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# 11. PLOT AVERAGE POSITION
# ============================================================

plt.figure(figsize=(8, 5))

plt.bar(
    comparison_df["Experiment"],
    comparison_df["Average Position"]
)

plt.ylabel("Average Finishing Position")
plt.title("Average Finishing Position Comparison")
plt.xticks(rotation=15)
plt.tight_layout()

position_plot = (
    "/kaggle/working/chefmind_position_comparison.png"
)

plt.savefig(
    position_plot,
    dpi=200,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# 12. SAVE RESULTS
# ============================================================

baseline_df.to_csv(
    "/kaggle/working/baseline_results.csv",
    index=False
)

shaped_df.to_csv(
    "/kaggle/working/shaped_results.csv",
    index=False
)

comparison_df.to_csv(
    "/kaggle/working/experiment_comparison.csv",
    index=False
)


# ============================================================
# 13. SAVE A SIMPLE SUMMARY
# ============================================================

summary_text = f"""
ChefMind-RL
Reinforcement Learning — Chef's Hat Environment

Experiment 1:
Official Dueling DQN baseline

Experiment 2:
ChefMind DQN with potential-based reward shaping

Baseline average win rate:
{baseline_win:.2f}%

ChefMind average win rate:
{shaped_win:.2f}%

Baseline average position:
{baseline_position:.2f}

ChefMind average position:
{shaped_position:.2f}

Baseline average reward:
{baseline_reward:.3f}

ChefMind average reward:
{shaped_reward:.3f}

Win-rate change:
{shaped_win - baseline_win:.2f} percentage points
"""

with open(
    "/kaggle/working/chefmind_summary.txt",
    "w"
) as f:
    f.write(summary_text)


print("\n")
print("=" * 70)
print("EXPERIMENT COMPLETE")
print("=" * 70)

print("Saved files:")
print("/kaggle/working/baseline_results.csv")
print("/kaggle/working/shaped_results.csv")
print("/kaggle/working/experiment_comparison.csv")
print("/kaggle/working/chefmind_win_rate_comparison.png")
print("/kaggle/working/chefmind_position_comparison.png")
print("/kaggle/working/chefmind_summary.txt")

```

```python
# ============================================================
# CHEFMIND-RL — FINAL TRAINING + EVALUATION PIPELINE
# ============================================================
# Variant: Sparse / Delayed Reward
# Algorithm: Dueling DQN
# Comparison:
#   1. Official DQN baseline
#   2. ChefMind DQN + potential-based reward shaping
#
# IMPORTANT:
# Previous 10-match run did not train because batch_size=512.
# This experiment uses batch_size=64.
# ============================================================

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from agents.agent_dqn import DQNAgent
from rooms.room import Room


# ============================================================
# CONFIGURATION
# ============================================================

NUM_MATCHES = 100
BATCH_SIZE = 64
MEMORY_SIZE = 10000

BASE_DIR = "/kaggle/working/ChefMind_Final"
BASELINE_DIR = os.path.join(BASE_DIR, "baseline")
SHAPED_DIR = os.path.join(BASE_DIR, "shaped")

os.makedirs(BASELINE_DIR, exist_ok=True)
os.makedirs(SHAPED_DIR, exist_ok=True)

print("=" * 75)
print("CHEFMIND-RL FINAL EXPERIMENT")
print("=" * 75)
print("Matches per experiment:", NUM_MATCHES)
print("Batch size:", BATCH_SIZE)
print("Memory size:", MEMORY_SIZE)
print("Algorithm: Dueling DQN")
print("=" * 75)


# ============================================================
# 1. CUSTOM CHEFMIND DQN
# ============================================================

class ChefMindDQNAgent(DQNAgent):
    """
    Custom DQN for the Sparse / Delayed Reward experiment.

    The official reward is retained.
    A small potential-based shaping signal is added
    according to progress in reducing cards in hand.
    """

    def _get_potential(self, state):

        # First 17 values represent the player's hand
        hand = state[:17]

        # Count remaining cards
        cards_remaining = np.count_nonzero(hand)

        # Fewer cards = better state
        return -float(cards_remaining)

    def _get_shaped_reward(
        self,
        old_state,
        new_state,
        base_reward
    ):

        old_potential = self._get_potential(old_state)
        new_potential = self._get_potential(new_state)

        shaping = (
            self.gamma * new_potential
            - old_potential
        )

        # Small coefficient prevents shaping from
        # dominating the original reward.
        return base_reward + 0.05 * shaping

    def request_action(self, observations):

        # -------------------------------
        # Convert observation to state
        # -------------------------------

        hand = (
            np.array(
                observations["hand"]
            ).flatten() / 13
        )

        board = (
            np.array(
                observations["board"]
            ).flatten() / 13
        )

        possible_actions_values = list(
            observations["possible_actions"]
        )

        obs = np.concatenate(
            [hand, board]
        )

        # -------------------------------
        # Legal action mask
        # -------------------------------

        possible_actions_mask = np.zeros(
            self.action_size,
            dtype=np.float32
        )

        valid_action_indices = [
            self.all_actions.index(value)
            for value in possible_actions_values
        ]

        possible_actions_mask[
            valid_action_indices
        ] = 1.0

        # -------------------------------
        # Select action
        # -------------------------------

        action_index = self.act(
            obs,
            possible_actions_mask,
            valid_action_indices
        )

        action_str = self.all_actions[
            action_index
        ]

        # -------------------------------
        # Original reward
        # -------------------------------

        base_reward = 0.0

        if action_str.lower() == "pass":
            base_reward -= 1.0

        base_reward -= 0.02

        # -------------------------------
        # Potential-based shaping
        # -------------------------------

        if (
            self.last_state is not None
            and self.last_action is not None
            and self.train
            and self.last_possible_actions is not None
        ):

            shaped_reward = (
                self._get_shaped_reward(
                    self.last_state,
                    obs,
                    base_reward
                )
            )

            self.episode.append(
                (
                    self.last_state,
                    self.last_possible_actions,
                    self.last_action,
                    shaped_reward,
                    obs,
                    possible_actions_mask,
                    False
                )
            )

        # Store current transition
        self.last_state = obs
        self.last_action = action_index
        self.last_possible_actions = (
            possible_actions_mask
        )

        return action_index


# ============================================================
# 2. CREATE BASELINE ROOM
# ============================================================

print("\nCreating baseline experiment...")

baseline_room = Room(
    run_remote_room=False,
    room_name="ChefMind_Final_Baseline",
    max_matches=NUM_MATCHES,
    output_folder=BASELINE_DIR,
    save_game_dataset=False,
    save_logs_game=False,
    save_logs_room=False
)


# ============================================================
# 3. CREATE BASELINE DQN AGENTS
# ============================================================

baseline_agents = []

for i in range(4):

    agent = DQNAgent(
        name=f"BaselineDQN{i}",
        state_size=28,
        action_size=200,
        gamma=0.99,
        lr=1e-4,
        epsilon=1.0,
        epsilon_min=0.05,
        epsilon_decay=0.995,
        batch_size=BATCH_SIZE,
        memory_size=MEMORY_SIZE,
        log_directory=baseline_room.room_dir,
        train=True,
        verbose_console=False
    )

    baseline_agents.append(agent)
    baseline_room.connect_player(agent)


print(
    "Baseline agents created:",
    len(baseline_agents)
)


# ============================================================
# 4. TRAIN BASELINE
# ============================================================

print("\n" + "=" * 75)
print("TRAINING BASELINE DQN")
print("=" * 75)

await baseline_room.run()

print("\nBaseline training completed!")


# ============================================================
# 5. CHECK WHETHER BASELINE ACTUALLY TRAINED
# ============================================================

print("\nBaseline training diagnostics:")

for agent in baseline_agents:

    print(
        f"{agent.name}: "
        f"loss={len(agent.loss_history)}, "
        f"epsilon_updates={len(agent.epsilon_history)}, "
        f"matches={len(agent.rewards)}"
    )


# ============================================================
# 6. CREATE SHAPED-REWARD ROOM
# ============================================================

print("\nCreating shaped-reward experiment...")

shaped_room = Room(
    run_remote_room=False,
    room_name="ChefMind_Final_Shaped",
    max_matches=NUM_MATCHES,
    output_folder=SHAPED_DIR,
    save_game_dataset=False,
    save_logs_game=False,
    save_logs_room=False
)


# ============================================================
# 7. CREATE SHAPED DQN AGENTS
# ============================================================

shaped_agents = []

for i in range(4):

    agent = ChefMindDQNAgent(
        name=f"ChefMindDQN{i}",
        state_size=28,
        action_size=200,
        gamma=0.99,
        lr=1e-4,
        epsilon=1.0,
        epsilon_min=0.05,
        epsilon_decay=0.995,
        batch_size=BATCH_SIZE,
        memory_size=MEMORY_SIZE,
        log_directory=shaped_room.room_dir,
        train=True,
        verbose_console=False
    )

    shaped_agents.append(agent)
    shaped_room.connect_player(agent)


print(
    "Shaped agents created:",
    len(shaped_agents)
)


# ============================================================
# 8. TRAIN CHEFMIND SHAPED DQN
# ============================================================

print("\n" + "=" * 75)
print("TRAINING CHEFMIND SHAPED DQN")
print("=" * 75)

await shaped_room.run()

print("\nChefMind shaped training completed!")


# ============================================================
# 9. CHECK WHETHER SHAPED DQN ACTUALLY TRAINED
# ============================================================

print("\nChefMind training diagnostics:")

for agent in shaped_agents:

    print(
        f"{agent.name}: "
        f"loss={len(agent.loss_history)}, "
        f"epsilon_updates={len(agent.epsilon_history)}, "
        f"matches={len(agent.rewards)}"
    )


# ============================================================
# 10. FUNCTION TO COLLECT RESULTS
# ============================================================

def collect_results(
    agents,
    experiment_name
):

    rows = []

    for agent in agents:

        rewards = np.array(
            agent.rewards,
            dtype=float
        )

        positions = np.array(
            agent.positions,
            dtype=float
        )

        wins = np.sum(
            positions == 1
        )

        matches = len(positions)

        win_rate = (
            wins / matches * 100
            if matches > 0
            else 0
        )

        average_position = (
            np.mean(positions)
            if matches > 0
            else np.nan
        )

        average_reward = (
            np.mean(rewards)
            if len(rewards) > 0
            else np.nan
        )

        rows.append({

            "Experiment":
                experiment_name,

            "Agent":
                agent.name,

            "Matches":
                matches,

            "Wins":
                int(wins),

            "Win Rate (%)":
                win_rate,

            "Average Position":
                average_position,

            "Average Reward":
                average_reward,

            "Training Updates":
                len(agent.loss_history),

            "Final Epsilon":
                (
                    agent.epsilon_history[-1]
                    if agent.epsilon_history
                    else agent.epsilon
                )
        })

    return pd.DataFrame(rows)


# ============================================================
# 11. COLLECT BOTH EXPERIMENT RESULTS
# ============================================================

baseline_results = collect_results(
    baseline_agents,
    "Baseline DQN"
)

shaped_results = collect_results(
    shaped_agents,
    "ChefMind Shaped DQN"
)

all_results = pd.concat(
    [
        baseline_results,
        shaped_results
    ],
    ignore_index=True
)


# ============================================================
# 12. DISPLAY PER-AGENT RESULTS
# ============================================================

print("\n" + "=" * 75)
print("FINAL PER-AGENT RESULTS")
print("=" * 75)

display(all_results)


# ============================================================
# 13. OVERALL EXPERIMENT COMPARISON
# ============================================================

overall_results = (
    all_results
    .groupby("Experiment")
    .agg({
        "Win Rate (%)": "mean",
        "Average Position": "mean",
        "Average Reward": "mean",
        "Training Updates": "mean"
    })
    .reset_index()
)

print("\n" + "=" * 75)
print("OVERALL COMPARISON")
print("=" * 75)

display(overall_results)


# ============================================================
# 14. CALCULATE IMPROVEMENTS
# ============================================================

base = overall_results[
    overall_results["Experiment"]
    == "Baseline DQN"
].iloc[0]

shape = overall_results[
    overall_results["Experiment"]
    == "ChefMind Shaped DQN"
].iloc[0]


win_change = (
    shape["Win Rate (%)"]
    - base["Win Rate (%)"]
)

position_change = (
    shape["Average Position"]
    - base["Average Position"]
)

reward_change = (
    shape["Average Reward"]
    - base["Average Reward"]
)


print("\n" + "=" * 75)
print("FINAL PERFORMANCE CHANGE")
print("=" * 75)

print(
    f"Win-rate change: "
    f"{win_change:+.2f} percentage points"
)

print(
    f"Average-position change: "
    f"{position_change:+.3f}"
)

print(
    f"Average-reward change: "
    f"{reward_change:+.4f}"
)


# ============================================================
# 15. SAVE RESULTS
# ============================================================

all_results.to_csv(
    os.path.join(
        BASE_DIR,
        "per_agent_results.csv"
    ),
    index=False
)

overall_results.to_csv(
    os.path.join(
        BASE_DIR,
        "overall_comparison.csv"
    ),
    index=False
)


# ============================================================
# 16. LEARNING CURVES — LOSS
# ============================================================

plt.figure(figsize=(10, 6))

for agent in baseline_agents:

    if len(agent.loss_history) > 0:

        plt.plot(
            agent.loss_history,
            label=agent.name
        )

plt.xlabel("Training Update")
plt.ylabel("Loss")
plt.title(
    "Baseline DQN Training Loss"
)

plt.legend()
plt.tight_layout()

baseline_loss_path = os.path.join(
    BASE_DIR,
    "baseline_loss.png"
)

plt.savefig(
    baseline_loss_path,
    dpi=200
)

plt.show()


plt.figure(figsize=(10, 6))

for agent in shaped_agents:

    if len(agent.loss_history) > 0:

        plt.plot(
            agent.loss_history,
            label=agent.name
        )

plt.xlabel("Training Update")
plt.ylabel("Loss")
plt.title(
    "ChefMind Shaped DQN Training Loss"
)

plt.legend()
plt.tight_layout()

shaped_loss_path = os.path.join(
    BASE_DIR,
    "shaped_loss.png"
)

plt.savefig(
    shaped_loss_path,
    dpi=200
)

plt.show()


# ============================================================
# 17. EPSILON CURVES
# ============================================================

plt.figure(figsize=(10, 6))

for agent in baseline_agents:

    if len(agent.epsilon_history) > 0:

        plt.plot(
            agent.epsilon_history,
            label=agent.name
        )

plt.xlabel("Training Update")
plt.ylabel("Epsilon")
plt.title(
    "Baseline DQN Exploration Decay"
)

plt.legend()
plt.tight_layout()

baseline_epsilon_path = os.path.join(
    BASE_DIR,
    "baseline_epsilon.png"
)

plt.savefig(
    baseline_epsilon_path,
    dpi=200
)

plt.show()


plt.figure(figsize=(10, 6))

for agent in shaped_agents:

    if len(agent.epsilon_history) > 0:

        plt.plot(
            agent.epsilon_history,
            label=agent.name
        )

plt.xlabel("Training Update")
plt.ylabel("Epsilon")
plt.title(
    "ChefMind Shaped DQN Exploration Decay"
)

plt.legend()
plt.tight_layout()

shaped_epsilon_path = os.path.join(
    BASE_DIR,
    "shaped_epsilon.png"
)

plt.savefig(
    shaped_epsilon_path,
    dpi=200
)

plt.show()


# ============================================================
# 18. WIN-RATE COMPARISON
# ============================================================

plt.figure(figsize=(9, 6))

plt.bar(
    overall_results["Experiment"],
    overall_results["Win Rate (%)"]
)

plt.ylabel("Average Win Rate (%)")
plt.title(
    "Baseline DQN vs ChefMind Shaped DQN"
)

plt.xticks(rotation=15)
plt.tight_layout()

win_path = os.path.join(
    BASE_DIR,
    "win_rate_comparison.png"
)

plt.savefig(
    win_path,
    dpi=200
)

plt.show()


# ============================================================
# 19. AVERAGE POSITION COMPARISON
# ============================================================

plt.figure(figsize=(9, 6))

plt.bar(
    overall_results["Experiment"],
    overall_results["Average Position"]
)

plt.ylabel("Average Finishing Position")
plt.title(
    "Average Finishing Position"
)

plt.xticks(rotation=15)
plt.tight_layout()

position_path = os.path.join(
    BASE_DIR,
    "average_position_comparison.png"
)

plt.savefig(
    position_path,
    dpi=200
)

plt.show()


# ============================================================
# 20. REWARD COMPARISON
# ============================================================

plt.figure(figsize=(9, 6))

plt.bar(
    overall_results["Experiment"],
    overall_results["Average Reward"]
)

plt.ylabel("Average Reward")
plt.title(
    "Average Reward Comparison"
)

plt.xticks(rotation=15)
plt.tight_layout()

reward_path = os.path.join(
    BASE_DIR,
    "average_reward_comparison.png"
)

plt.savefig(
    reward_path,
    dpi=200
)

plt.show()


# ============================================================
# 21. SAVE EXPERIMENT SUMMARY
# ============================================================

summary = f"""
============================================================
ChefMind-RL Final Experiment
============================================================

Environment:
Official Chef's Hat environment

Algorithm:
Dueling Deep Q-Network (DQN)

Variant:
Sparse / Delayed Reward

Number of agents:
4

Matches per experiment:
{NUM_MATCHES}

Batch size:
{BATCH_SIZE}

------------------------------------------------------------
BASELINE DQN
------------------------------------------------------------

Average win rate:
{base["Win Rate (%)"]:.2f}%

Average finishing position:
{base["Average Position"]:.3f}

Average reward:
{base["Average Reward"]:.4f}

Average training updates:
{base["Training Updates"]:.2f}

------------------------------------------------------------
CHEFMIND SHAPED DQN
------------------------------------------------------------

Average win rate:
{shape["Win Rate (%)"]:.2f}%

Average finishing position:
{shape["Average Position"]:.3f}

Average reward:
{shape["Average Reward"]:.4f}

Average training updates:
{shape["Training Updates"]:.2f}

------------------------------------------------------------
CHANGE
------------------------------------------------------------

Win-rate change:
{win_change:+.2f} percentage points

Average-position change:
{position_change:+.3f}

Average-reward change:
{reward_change:+.4f}

------------------------------------------------------------

Interpretation:

The final conclusion must be based on the observed
experimental results. No performance improvement should
be claimed unless supported by the measured results.
"""

summary_path = os.path.join(
    BASE_DIR,
    "final_summary.txt"
)

with open(
    summary_path,
    "w"
) as file:

    file.write(summary)


# ============================================================
# 22. FINAL FILE LIST
# ============================================================

print("\n" + "=" * 75)
print("CHEFMIND FINAL EXPERIMENT COMPLETE")
print("=" * 75)

print("\nResults saved in:")
print(BASE_DIR)

print("\nImportant files:")

for filename in sorted(
    os.listdir(BASE_DIR)
):

    print(
        " -",
        filename
    )


# ============================================================
# 23. FINAL DIAGNOSTIC
# ============================================================

print("\n" + "=" * 75)
print("TRAINING VERIFICATION")
print("=" * 75)

for agent in baseline_agents:

    print(
        f"{agent.name}: "
        f"{len(agent.loss_history)} training updates"
    )

for agent in shaped_agents:

    print(
        f"{agent.name}: "
        f"{len(agent.loss_history)} training updates"
    )

print("\nIf the training-update values are greater than 0,")
print("the DQN replay/training mechanism was actually used.")
print("=" * 75)
```

```python
# ============================================================
# CHEFMIND-RL — FINAL EVALUATION
# ============================================================
# Purpose:
# Evaluate the trained Baseline DQN and ChefMind Shaped DQN
# using greedy action selection (epsilon = 0).
# ============================================================

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

EVAL_MATCHES = 100

EVAL_DIR = "/kaggle/working/ChefMind_Final/evaluation"
BASELINE_EVAL_DIR = os.path.join(EVAL_DIR, "baseline")
SHAPED_EVAL_DIR = os.path.join(EVAL_DIR, "shaped")

os.makedirs(BASELINE_EVAL_DIR, exist_ok=True)
os.makedirs(SHAPED_EVAL_DIR, exist_ok=True)

print("=" * 75)
print("CHEFMIND-RL — GREEDY EVALUATION")
print("=" * 75)
print("Evaluation matches:", EVAL_MATCHES)
print("Exploration epsilon: 0.0")
print("=" * 75)


# ------------------------------------------------------------
# Create fresh evaluation agents
# ------------------------------------------------------------

baseline_eval_room = Room(
    run_remote_room=False,
    room_name="ChefMind_Evaluation_Baseline",
    max_matches=EVAL_MATCHES,
    output_folder=BASELINE_EVAL_DIR,
    save_game_dataset=False,
    save_logs_game=False,
    save_logs_room=False
)

shaped_eval_room = Room(
    run_remote_room=False,
    room_name="ChefMind_Evaluation_Shaped",
    max_matches=EVAL_MATCHES,
    output_folder=SHAPED_EVAL_DIR,
    save_game_dataset=False,
    save_logs_game=False,
    save_logs_room=False
)


baseline_eval_agents = []
shaped_eval_agents = []


# ------------------------------------------------------------
# IMPORTANT:
# Copy trained neural-network weights into fresh agents.
# Set epsilon = 0 for greedy evaluation.
# ------------------------------------------------------------

for i, trained_agent in enumerate(baseline_agents):

    eval_agent = DQNAgent(
        name=f"BaselineEval{i}",
        state_size=28,
        action_size=200,
        gamma=0.99,
        lr=1e-4,
        epsilon=0.0,
        epsilon_min=0.0,
        epsilon_decay=1.0,
        batch_size=64,
        memory_size=10000,
        log_directory=baseline_eval_room.room_dir,
        train=False,
        verbose_console=False
    )

    eval_agent.model.set_weights(
        trained_agent.model.get_weights()
    )

    eval_agent.target_model.set_weights(
        trained_agent.target_model.get_weights()
    )

    baseline_eval_agents.append(eval_agent)
    baseline_eval_room.connect_player(eval_agent)


for i, trained_agent in enumerate(shaped_agents):

    eval_agent = ChefMindDQNAgent(
        name=f"ChefMindEval{i}",
        state_size=28,
        action_size=200,
        gamma=0.99,
        lr=1e-4,
        epsilon=0.0,
        epsilon_min=0.0,
        epsilon_decay=1.0,
        batch_size=64,
        memory_size=10000,
        log_directory=shaped_eval_room.room_dir,
        train=False,
        verbose_console=False
    )

    eval_agent.model.set_weights(
        trained_agent.model.get_weights()
    )

    eval_agent.target_model.set_weights(
        trained_agent.target_model.get_weights()
    )

    shaped_eval_agents.append(eval_agent)
    shaped_eval_room.connect_player(eval_agent)


print("\nEvaluation agents created successfully.")


# ------------------------------------------------------------
# Run baseline evaluation
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("EVALUATING BASELINE DQN")
print("=" * 75)

await baseline_eval_room.run()

print("\nBaseline evaluation completed.")


# ------------------------------------------------------------
# Run shaped evaluation
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("EVALUATING CHEFMIND SHAPED DQN")
print("=" * 75)

await shaped_eval_room.run()

print("\nChefMind evaluation completed.")


# ------------------------------------------------------------
# Collect evaluation results
# ------------------------------------------------------------

def collect_evaluation_results(agents, experiment):

    rows = []

    for agent in agents:

        rewards = np.array(agent.rewards, dtype=float)
        positions = np.array(agent.positions, dtype=float)

        matches = len(positions)

        wins = np.sum(positions == 1)

        win_rate = (
            wins / matches * 100
            if matches > 0
            else 0
        )

        average_position = (
            np.mean(positions)
            if matches > 0
            else np.nan
        )

        average_reward = (
            np.mean(rewards)
            if len(rewards) > 0
            else np.nan
        )

        rows.append({
            "Experiment": experiment,
            "Agent": agent.name,
            "Matches": matches,
            "Wins": int(wins),
            "Win Rate (%)": win_rate,
            "Average Position": average_position,
            "Average Reward": average_reward
        })

    return pd.DataFrame(rows)


baseline_eval_results = collect_evaluation_results(
    baseline_eval_agents,
    "Baseline DQN"
)

shaped_eval_results = collect_evaluation_results(
    shaped_eval_agents,
    "ChefMind Shaped DQN"
)


evaluation_results = pd.concat(
    [
        baseline_eval_results,
        shaped_eval_results
    ],
    ignore_index=True
)


print("\n" + "=" * 75)
print("PER-AGENT EVALUATION RESULTS")
print("=" * 75)

display(evaluation_results)


# ------------------------------------------------------------
# Overall evaluation comparison
# ------------------------------------------------------------

evaluation_overall = (
    evaluation_results
    .groupby("Experiment")
    .agg({
        "Win Rate (%)": "mean",
        "Average Position": "mean",
        "Average Reward": "mean"
    })
    .reset_index()
)


print("\n" + "=" * 75)
print("OVERALL EVALUATION")
print("=" * 75)

display(evaluation_overall)


# ------------------------------------------------------------
# Calculate performance differences
# ------------------------------------------------------------

base_eval = evaluation_overall[
    evaluation_overall["Experiment"] == "Baseline DQN"
].iloc[0]

shape_eval = evaluation_overall[
    evaluation_overall["Experiment"] == "ChefMind Shaped DQN"
].iloc[0]


win_rate_change = (
    shape_eval["Win Rate (%)"]
    - base_eval["Win Rate (%)"]
)

position_change = (
    shape_eval["Average Position"]
    - base_eval["Average Position"]
)

reward_change = (
    shape_eval["Average Reward"]
    - base_eval["Average Reward"]
)


print("\n" + "=" * 75)
print("EVALUATION PERFORMANCE CHANGE")
print("=" * 75)

print(
    f"Win-rate change: "
    f"{win_rate_change:+.2f} percentage points"
)

print(
    f"Average-position change: "
    f"{position_change:+.3f}"
)

print(
    f"Average-reward change: "
    f"{reward_change:+.4f}"
)


# ------------------------------------------------------------
# Save evaluation tables
# ------------------------------------------------------------

evaluation_results.to_csv(
    os.path.join(
        EVAL_DIR,
        "evaluation_per_agent.csv"
    ),
    index=False
)

evaluation_overall.to_csv(
    os.path.join(
        EVAL_DIR,
        "evaluation_overall.csv"
    ),
    index=False
)


# ------------------------------------------------------------
# Evaluation plots
# ------------------------------------------------------------

plt.figure(figsize=(9, 6))

plt.bar(
    evaluation_overall["Experiment"],
    evaluation_overall["Win Rate (%)"]
)

plt.ylabel("Win Rate (%)")
plt.title("Greedy Evaluation: Win Rate")
plt.xticks(rotation=15)
plt.tight_layout()

plt.savefig(
    os.path.join(
        EVAL_DIR,
        "evaluation_win_rate.png"
    ),
    dpi=200
)

plt.show()


plt.figure(figsize=(9, 6))

plt.bar(
    evaluation_overall["Experiment"],
    evaluation_overall["Average Position"]
)

plt.ylabel("Average Finishing Position")
plt.title("Greedy Evaluation: Average Position")
plt.xticks(rotation=15)
plt.tight_layout()

plt.savefig(
    os.path.join(
        EVAL_DIR,
        "evaluation_average_position.png"
    ),
    dpi=200
)

plt.show()


plt.figure(figsize=(9, 6))

plt.bar(
    evaluation_overall["Experiment"],
    evaluation_overall["Average Reward"]
)

plt.ylabel("Average Reward")
plt.title("Greedy Evaluation: Average Reward")
plt.xticks(rotation=15)
plt.tight_layout()

plt.savefig(
    os.path.join(
        EVAL_DIR,
        "evaluation_average_reward.png"
    ),
    dpi=200
)

plt.show()


# ------------------------------------------------------------
# Final evaluation summary
# ------------------------------------------------------------

evaluation_summary = f"""
============================================================
ChefMind-RL — Greedy Evaluation
============================================================

Evaluation matches:
{EVAL_MATCHES}

Exploration:
epsilon = 0

------------------------------------------------------------
BASELINE DQN
------------------------------------------------------------

Win rate:
{base_eval["Win Rate (%)"]:.2f}%

Average finishing position:
{base_eval["Average Position"]:.3f}

Average reward:
{base_eval["Average Reward"]:.4f}

------------------------------------------------------------
CHEFMIND SHAPED DQN
------------------------------------------------------------

Win rate:
{shape_eval["Win Rate (%)"]:.2f}%

Average finishing position:
{shape_eval["Average Position"]:.3f}

Average reward:
{shape_eval["Average Reward"]:.4f}

------------------------------------------------------------
CHANGE
------------------------------------------------------------

Win-rate change:
{win_rate_change:+.2f} percentage points

Average-position change:
{position_change:+.3f}

Average-reward change:
{reward_change:+.4f}

------------------------------------------------------------
INTERPRETATION
------------------------------------------------------------

The evaluation compares the learned policies under
greedy action selection.

A positive win-rate change indicates better winning
performance for the shaped agent.

A negative average-position change indicates better
finishing position because position 1 is the best.

The final project conclusion must be based on the
observed evaluation results.
"""

with open(
    os.path.join(
        EVAL_DIR,
        "evaluation_summary.txt"
    ),
    "w"
) as f:
    f.write(evaluation_summary)


print("\n" + "=" * 75)
print("GREEDY EVALUATION COMPLETE")
print("=" * 75)

print("\nEvaluation files saved to:")
print(EVAL_DIR)

for filename in sorted(os.listdir(EVAL_DIR)):
    print(" -", filename)
```

## Conclusion

This project implemented a Dueling Deep Q-Network in the official
Chef's Hat environment for the Sparse / Delayed Reward variant.

A baseline DQN was compared with a DQN incorporating
potential-based reward shaping.

Both approaches achieved an average win rate of 25.0%, an average
finishing position of 2.50 and an average reward of 0.735.

Therefore, reward shaping did not provide a measurable overall
performance improvement under the tested configuration.

However, the individual agents behaved differently. The best
ChefMind shaped agent achieved a 37% win rate and an average
finishing position of 2.05, compared with 33% and 2.18 for the
best baseline agent.

The results therefore suggest that reward shaping influenced
learning behaviour but did not provide a consistent aggregate
advantage in this experiment.
